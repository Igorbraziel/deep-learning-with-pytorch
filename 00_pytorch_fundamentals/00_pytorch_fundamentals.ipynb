## Reshaping, Viewing and Stacking

```python
y = torch.arange(1, 41)
y
```

```python
y_reshaped = y.reshape(10, 4)
y_reshaped
```

```python
z = y.view(4, 10) # It creates a copy of the tensor with a different shape, but sharing the same memory
z
```

```python
z[:, 0] = 100 # It changes the original tensor 'y'
z, y
```

```python
z_stacked = torch.stack([z, z, z], dim=0)
z_stacked
```

```python
z_stacked = torch.stack([z, z, z], dim=1)
z_stacked
```

## Squeezing, Unsqueezing and Permuting

## GPU access with PyTorch

```python
torch.cuda.is_available()
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
torch.cuda.device_count()
```

```python
tensor_on_cpu = torch.arange(0, 10)
tensor_on_cpu.device
```

```python
tensor_on_gpu = tensor_on_cpu.to(device)
tensor_on_gpu.device
```

```python
backing_to_cpu = tensor_on_gpu.cpu()
backing_to_cpu.device
```
