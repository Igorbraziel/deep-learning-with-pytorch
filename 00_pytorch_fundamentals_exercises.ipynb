### 4. Set the random seed to `0` and do 2 & 3 over again.

The output should be:
```
(tensor([[1.8542],
         [1.9611],
         [2.2884],
         [3.0481],
         [1.7067],
         [2.5290],
         [1.7989]]), torch.Size([7, 1]))
```

````python
RANDOM_SEED = 0

torch.manual_seed(RANDOM_SEED)
tensor_1 = torch.rand(7, 7)

torch.manual_seed(RANDOM_SEED)
tensor_2 = torch.rand(1, 7)

tensor_2_reshaped = tensor_2.reshape(7, 1)

result = torch.matmul(tensor_1, tensor_2_reshaped)
result, result.shape


````


### 6. Create two random tensors of shape `(2, 3)` and send them both to the GPU (you'll need access to a GPU for this). Set `torch.manual_seed(1234)` when creating the tensors (this doesn't have to be the GPU random seed). The output should be something like:


````python
device = "cuda" if torch.cuda.is_available() else "cpu"

torch.manual_seed(GPU_RANDOM_SEED)
x = torch.rand(2, 3, device=device)

torch.manual_seed(GPU_RANDOM_SEED)
y = torch.rand(2, 3, device=device)

device, x, y
````
